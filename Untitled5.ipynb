```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1️⃣ قراءة الملف
df = pd.read_excel("Book2.xlsx")

# 2️⃣ تحويل العمود اللي فيه كل البيانات
df = df["Data"].str.split(",", expand=True)

# 3️⃣ حذف أول صف لأنه Header غلط داخل الداتا
df = df.iloc[1:].reset_index(drop=True)

# 4️⃣ تسمية الأعمدة
df.columns = ["Date", "Category", "Amount", "Payment_Method"]

# 5️⃣ تنظيف وتحويل أنواع البيانات
df["Amount"] = pd.to_numeric(df["Amount"], errors="coerce")
df["Date"] = pd.to_datetime(df["Date"], errors="coerce")

# 6️⃣ حذف أي قيم خربانة
df = df.dropna()

# =========================================
# 📊 Analysis
# =========================================

print("Total Spending:", df["Amount"].sum())

category_spending = df.groupby("Category")["Amount"].sum()
print("\nSpending by Category:")
print(category_spending)

top_category = category_spending.idxmax()
print("\nTop Category:", top_category)

# =========================================
# 📈 Visualization
# =========================================

plt.figure(figsize=(6,6))
category_spending.plot.pie(autopct='%1.1f%%')
plt.title("Spending Distribution")
plt.ylabel("")
plt.show()

plt.figure(figsize=(8,5))
sns.barplot(x=category_spending.index, y=category_spending.values)
plt.show()

daily = df.groupby("Date")["Amount"].sum()

plt.figure(figsize=(8,5))
sns.lineplot(x=daily.index, y=daily.values)
plt.show()

# =========================================
# 💡 Insight
# =========================================

print("\nInsight:")
print("Most spending category:", top_category)
print("\nFinal Insight:")
print("Most of the spending is concentrated in Shopping, indicating impulsive or non-essential purchases dominate expenses.")
```
