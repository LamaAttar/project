```python
import pandas as pd
import matplotlib.pyplot as plt

# =========================
# ⚽ Data (Simple but stronger logic)
# =========================

data = {
    "Team": ["Jordan 🇯🇴", "Austria 🇦🇹"],
    "Goals_Scored": [8, 15],
    "Goals_Conceded": [6, 7],
    "Matches": [5, 5],
    "FIFA_Rank_Score": [55, 78]  # added feature for realism
}

df = pd.DataFrame(data)

# =========================
# 🧠 Feature Engineering
# =========================

df["Attack_Strength"] = df["Goals_Scored"] / df["Matches"]
df["Defense_Strength"] = 1 / (df["Goals_Conceded"] / df["Matches"])

# Weighted score (more realistic)
df["Overall_Score"] = (
    (df["Attack_Strength"] * 0.5) +
    (df["Defense_Strength"] * 0.3) +
    (df["FIFA_Rank_Score"] * 0.2)
)

print(df)

# =========================
# 🔥 Prediction
# =========================

jordan_score = df.loc[0, "Overall_Score"]
austria_score = df.loc[1, "Overall_Score"]

total = jordan_score + austria_score

jordan_prob = round((jordan_score / total) * 100, 2)
austria_prob = round((austria_score / total) * 100, 2)

print("\n🎯 Match Prediction Probabilities:")
print("Jordan 🇯🇴:", jordan_prob, "%")
print("Austria 🇦🇹:", austria_prob, "%")

winner = "Jordan 🇯🇴" if jordan_prob > austria_prob else "Austria 🇦🇹"
print("\n🏆 Predicted Winner:", winner)

# =========================
# 📊 Visualization (VERY important for LinkedIn)
# =========================

plt.figure(figsize=(6,4))
plt.bar(["Jordan 🇯🇴", "Austria 🇦🇹"], [jordan_prob, austria_prob])
plt.title("Win Probability Prediction")
plt.ylabel("Probability (%)")
plt.show()
```
